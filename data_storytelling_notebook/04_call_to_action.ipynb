## 🎯 Three-Tier Takeaways: "So What?" Answered for Three Audiences

### **Tier 1: For Practitioners (Tactical Actions)**

Here are **3 specific actions** you can apply to your next fraud detection project:

**1️⃣  PRIORITIZE EDA BEFORE MODELING**
- Spend 50% of project time understanding data distribution
- Identify power-law patterns, class imbalance, outliers
- Don't trust "high accuracy" on imbalanced data (like our 50.09% baseline that caught 0 phishers!)
- **Tool**: Use log-log plots to detect power-law distributions

**2️⃣  ENGINEER BEHAVIORAL FEATURES, NOT JUST AGGREGATES**
- Go beyond counts/sums capture temporal patterns (short-term vs long-term frequency)
- Compare account behavior across dimensions (direction ratio, time-of-day patterns)
- Network features (centrality measures) reveal hidden structure
- **Example**: Our `direction_ratio` (incoming/outgoing) was the #1 predictor (ρ=0.426)

**3️⃣  SELECT FEATURES WITH ROBUST METHODS**
- Use Spearman (rank-based) for non-normal data, NOT Pearson
- Validate across multiple class distributions (balanced, imbalanced, real-world)
- Test on truly held-out data with realistic imbalance ratios
- **Result**: We went from 34→17 features while maintaining 92.7% recall
